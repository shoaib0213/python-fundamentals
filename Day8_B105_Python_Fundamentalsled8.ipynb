```python
useraccount = {'username':"codetraining",'firstname':'code','lastname':"training",'dob':"01-01-2020",'pwd':'54321'}
```

```python
for a,b in useraccount.items():
    print(a)
    print(b)
```

```python
# enhance of the code----
```

```python
for a,b in useraccount.items():
    print(f"key:{a}")
    print(f"values:{b}")
```

```python
for a,b in useraccount.items():
    print(f"key:{a}")
    print(f"value:{b}\n")
```

```python
# i want to get only the keys----
```

```python
for a in useraccount.keys():
    print(a)
```

```python
for a in useraccount.keys():
    print(f"keys : {a}")
```

```python
# i want to get only the values----
```

```python
for b in useraccount.values():
    print(f"values:{b}")
```
